```python

import pandas as pd

data = {
    "Date": ["2026-01-01", "2026-01-02", "2026-01-03", "2026-01-04", "2026-01-05",
             "2026-01-06", "2026-01-07", "2026-01-08", "2026-01-09", "2026-01-10"],
    "Region": ["North", "South ", "East", "West", "North",
               "South", "East ", "West", "North", "South"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile",
                "Tablet", "Laptop", "Mobile", "Tablet", "Laptop"],
    "Quantity": [2, 3, 1, 4, 2, None, 3, 2, 1, 4],
    "Sales": [100000, 45000, 20000, 200000, 50000,
              30000, 120000, None, 25000, 180000],
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)

print("Dataset created successfully!")
df.head()
```

```python

import pandas as pd

df = pd.read_csv("sales_data.csv")

print("Dataset loaded successfully!")
```

```python

# Display first 5 rows
display(df.head())

# Check dataset size
print("Shape:", df.shape)

# Check column names and data types
df.info()

# Statistical summary
display(df.describe())
```

```python

# 1. Check missing values
print("Missing values:")
print(df.isnull().sum())

# 2. Check duplicate records
print("\nDuplicate rows:", df.duplicated().sum())

# 3. Check data types
print("\nData types:")
print(df.dtypes)

# 4. Check extra spaces in text columns
print("\nRegion values:")
print(df["Region"].unique())

# 5. Check unique values in Product
print("\nProduct values:")
print(df["Product"].unique())

# 6. Check invalid or negative quantities
print("\nInvalid quantities:")
print(df[df["Quantity"] < 0])
```

```python

# Fill missing Quantity values with median
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())

# Fill missing Sales values with median
df["Sales"] = df["Sales"].fillna(df["Sales"].median())

print("Missing values filled successfully!")
print(df.isnull().sum())
```

```python

# Remove extra spaces from text columns
df["Region"] = df["Region"].str.strip()
df["Product"] = df["Product"].str.strip()
df["Customer"] = df["Customer"].str.strip()

print(df["Region"].unique())
```

```python

df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
```

```python

df = df.drop_duplicates()

print("Duplicate rows removed!")
print("Shape after cleaning:", df.shape)
```

```python

# Calculate total sales for each region
region_sales = df.groupby("Region")["Sales"].sum()

print("Region-wise Total Sales:")
display(region_sales)
```

```python

# Calculate average sales for each product
product_sales = df.groupby("Product")["Sales"].mean()

print("Product-wise Average Sales:")
display(product_sales)
```

```python

region_analysis = df.groupby("Region").agg({
    "Sales": "sum",
    "Quantity": "sum"
})

display(region_analysis)
```

```python

# Create a pivot table
pivot_table = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print("Pivot Table: Region-wise and Product-wise Sales")
display(pivot_table)
```

```python

print("Total Sales:", df["Sales"].sum())
print("Total Quantity:", df["Quantity"].sum())
print("Average Sales:", df["Sales"].mean())
print("Total Records:", len(df))
```

### Data Quality Summary

1. I filled the empty values in Quantity and Sales using the median.
2. I removed extra spaces from the text values.
3. I changed the Date column into the correct date format.
4. I checked for duplicate rows and removed them.
5. I used GroupBy and a pivot table to understand the sales data better.
